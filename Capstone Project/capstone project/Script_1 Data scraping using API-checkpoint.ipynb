```python
"""
========================================================
NIFTY 50 Stock Market Analysis - Capstone Project
Script 1: Data Acquisition via Yahoo Finance API
========================================================
Author   : Data Science Consultant
Domain   : Financial Markets / Stock Analysis
Data Src : Yahoo Finance API (yfinance)

========================================================

Business Problem:
-----------------
Indian equity markets, specifically the NIFTY 50 index constituents, hold
immense data that retail and institutional investors rely on for decision
making. This project aims to:
  1. Collect historical OHLCV (Open, High, Low, Close, Volume) data for
     15 major NIFTY 50 stocks spanning 5 years (2020-2024).
  2. Clean, engineer features, and analyse price trends and volatility.
  3. Build ML models to predict next-day price direction (Classification)
     and cluster stocks by risk-return profile (Clustering).
  4. Communicate findings via a dashboard / visualisation report.

This script (01) handles DATA ACQUISITION only.
Run 02_cleaning_modelling.py for the rest of the pipeline.
"""

import pandas as pd
import datetime as dt
import os, time, requests, io

# ─────────────────────────────────────────────
# CONFIGURATION
# ─────────────────────────────────────────────
SYMBOLS = [
    "RELIANCE.NS", "TCS.NS", "INFY.NS", "HDFCBANK.NS", "ICICIBANK.NS",
    "SBIN.NS", "LT.NS", "ITC.NS", "KOTAKBANK.NS", "AXISBANK.NS",
    "HINDUNILVR.NS", "BHARTIARTL.NS", "ASIANPAINT.NS", "BAJFINANCE.NS",
    "MARUTI.NS"
]

START_DATE = "2020-01-01"
END_DATE   = "2024-12-31"
OUTPUT_CSV = "nifty50_yahoo_api_data.csv"

# ─────────────────────────────────────────────
# HELPER: Download via Yahoo Finance v8 API
# ─────────────────────────────────────────────

def fetch_yahoo_data(symbol: str, start: str, end: str) -> pd.DataFrame:
    """
    Fetches historical OHLCV data from Yahoo Finance public API.
    No login required – uses the publicly accessible /v8/finance/chart endpoint.

    Parameters
    ----------
    symbol : str   Ticker symbol e.g. 'RELIANCE.NS'
    start  : str   Start date 'YYYY-MM-DD'
    end    : str   End date   'YYYY-MM-DD'

    Returns
    -------
    pd.DataFrame with columns: Date, Open, High, Low, Close, Volume, Symbol
    """
    start_ts = int(dt.datetime.strptime(start, "%Y-%m-%d").timestamp())
    end_ts   = int(dt.datetime.strptime(end,   "%Y-%m-%d").timestamp())

    url = (
        f"https://query1.finance.yahoo.com/v8/finance/chart/{symbol}"
        f"?period1={start_ts}&period2={end_ts}&interval=1d"
        f"&includeAdjustedClose=true"
    )
    headers = {"User-Agent": "Mozilla/5.0"}

    try:
        r = requests.get(url, headers=headers, timeout=15)
        r.raise_for_status()
        data = r.json()

        result = data["chart"]["result"][0]
        timestamps = result["timestamp"]
        q = result["indicators"]["quote"][0]

        df = pd.DataFrame({
            "Date"  : pd.to_datetime(timestamps, unit="s"),
            "Open"  : q["open"],
            "High"  : q["high"],
            "Low"   : q["low"],
            "Close" : q["close"],
            "Volume": q["volume"],
        })
        df["Symbol"] = symbol
        df.dropna(inplace=True)
        return df

    except Exception as e:
        print(f"  [ERROR] {symbol}: {e}")
        return pd.DataFrame()


# ─────────────────────────────────────────────
# MAIN SCRAPING LOOP
# ─────────────────────────────────────────────

def main():
    print("=" * 60)
    print("  NIFTY 50 – Yahoo Finance Data Scraper")
    print(f"  Symbols : {len(SYMBOLS)}")
    print(f"  Period  : {START_DATE}  →  {END_DATE}")
    print("=" * 60)

    all_frames = []

    for i, sym in enumerate(SYMBOLS, 1):
        print(f"  [{i:02d}/{len(SYMBOLS)}] Fetching {sym} ...", end=" ")
        df = fetch_yahoo_data(sym, START_DATE, END_DATE)
        if not df.empty:
            print(f"{len(df)} rows ✓")
            all_frames.append(df)
        else:
            print("SKIPPED")
        time.sleep(0.5)   # polite delay – avoid rate limiting

    if not all_frames:
        print("\n[!] No data fetched. Check network / symbols.")
        return

    combined = pd.concat(all_frames, ignore_index=True)
    combined.sort_values(["Symbol", "Date"], inplace=True)
    combined.reset_index(drop=True, inplace=True)

    # Basic validation
    print(f"\n  Total records   : {len(combined):,}")
    print(f"  Unique symbols  : {combined['Symbol'].nunique()}")
    print(f"  Date range      : {combined['Date'].min().date()} → {combined['Date'].max().date()}")
    print(f"  Missing values  : {combined.isnull().sum().sum()}")

    combined.to_csv(OUTPUT_CSV, index=False)
    print(f"\n  ✅  Data saved to '{OUTPUT_CSV}'")
    print("=" * 60)


if __name__ == "__main__":
    # ── If already downloaded (e.g. the provided CSV), skip re-scraping ──
    if os.path.exists("nifty50_yahoo_api_data.csv"):
        print(f"[INFO] '{OUTPUT_CSV}' already exists. Delete it to re-scrape.")
        df = pd.read_csv("nifty50_yahoo_api_data.csv")
        print(f"[INFO] Loaded {len(df):,} rows from existing file.")
    else:
        main()
```
